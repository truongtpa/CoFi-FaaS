# Cost–Time Plan Optimization

Plans estimated by the optimizer (`cofi/plans.py`) from the measured runs, with the v5 cost-time model calibrated
leave-one-query-out. Knobs per query: which Bloom filters are pushed down, and `max_size_mb` ∈ {10, 25, 50, 100,
200, 400, 1000} of every scan / aggregate / broadcast-join stage (functions reading more than 200 MB are excluded).
Coordinate descent on λ·T/T₀ + (1−λ)·C/C₀, λ ∈ {0, 0.1, …, 1}; the knee is the Pareto point closest to
(T_min, C_min) after normalisation. All values are **estimates** (see section 4 for measured plans).

| File | Content |
|---|---|
| `plans.csv` | per query: `default` (all BFs on, 50 MB), `knee`, `min_T`, `min_C`, `front_01`…`front_06` (`run_as`: the plan actually run when two are equal), with estimated T (s) and C ($), BFs turned off, sizes changed, functions per stage |
| `points.csv` | every plan evaluated by the search (T, C, on the Pareto front, knee, default) |
| `p-check.csv` | functions per stage simulated at the default size vs. measured |
| `bf-tradeoff.csv` | per query: estimated T / C with no BF, default, BF build part, saving, knee |
| `validation-runs.csv` | measured runs of the exported plans (part 2) |

```python
import os, sys, json
sys.path.insert(0, os.path.abspath(".."))      # repository root: cofi/style.py
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from cofi import style
%matplotlib inline
style.apply()
FONT_SIZE = style.FONT_SIZE
DATA, FIG = "data", "figures"
os.makedirs(FIG, exist_ok=True)
pd.set_option("display.width", 160)
qnum = lambda q: int("".join(c for c in q if c.isdigit()) or 0)

PL = pd.read_csv(f"{DATA}/plans.csv")
PTS = pd.read_csv(f"{DATA}/points.csv")
CHK = pd.read_csv(f"{DATA}/p-check.csv")
PTS["C"] = PTS["C"] * style.COST_SCALE
PLANS = {(r.query, r.plan): r for r in PL.itertuples()}
QS = sorted(PL["query"].unique(), key=qnum)
print(f"functions per stage, simulated == measured at the default size: {(CHK.p_meas == CHK.p_sim).sum()}/{len(CHK)}")
```

## Default plan, Bloom-filter build cost and saving (estimated)

Per query, with the same model and calibration as the optimizer:
* **No BF** — every Bloom filter off, default sizes; **Default** — every Bloom filter on (BF always on, 50 MB);
* **BF build** — what the Bloom filters cost inside the default plan: the merge_bf stages, plus the BF part of the
  other functions (hashing to build / apply the filter, reading / writing BF files);
* **Saving (net)** = No BF − Default; **saving (gross)** = net + BF build, i.e. what the smaller data downstream of
  the filters saves before paying for the filters;
* **Knee** — the plan selected by the optimizer.

```python
BT = pd.read_csv(f"{DATA}/bf-tradeoff.csv").set_index("query").reindex(QS)
S = style.COST_SCALE
show = pd.DataFrame({
    "BFs": BT["n_bf"],
    "T no BF (s)": BT["T_no_bf_s"], "T default (s)": BT["T_default_s"], "T BF build (s)": BT["T_bf_build_s"],
    "T saving net (s)": BT["T_saving_net_s"], "T knee (s)": BT["T_knee_s"],
    f"C no BF ({style.COST_UNIT})": BT["C_no_bf_usd"] * S, f"C default ({style.COST_UNIT})": BT["C_default_usd"] * S,
    f"C BF build ({style.COST_UNIT})": BT["C_bf_build_usd"] * S, f"C saving net ({style.COST_UNIT})": BT["C_saving_net_usd"] * S,
    f"C knee ({style.COST_UNIT})": BT["C_knee_usd"] * S,
})
show.loc["total"] = show.sum()
show.index = [q.upper() for q in show.index]
show.round(3)
```

```python
tot = BT.sum(numeric_only=True)
summ = pd.DataFrame({
    "No BF": [tot["T_no_bf_s"], tot["C_no_bf_usd"] * S],
    "Default (BF always on)": [tot["T_default_s"], tot["C_default_usd"] * S],
    "of which BF build": [tot["T_bf_build_s"], tot["C_bf_build_usd"] * S],
    "Saving gross (data reduced by the BFs)": [tot["T_saving_gross_s"], tot["C_saving_gross_usd"] * S],
    "Saving net = No BF − Default": [tot["T_saving_net_s"], tot["C_saving_net_usd"] * S],
    "Knee (optimized)": [tot["T_knee_s"], tot["C_knee_usd"] * S],
}, index=["T over 19 queries (s)", f"C over 19 queries ({style.COST_UNIT})"]).T
summ["ΔT vs default"] = (summ.iloc[:, 0] / summ.loc["Default (BF always on)"].iloc[0] - 1).map("{:+.0%}".format)
summ["ΔC vs default"] = (summ.iloc[:, 1] / summ.loc["Default (BF always on)"].iloc[1] - 1).map("{:+.0%}".format)
summ.loc[["of which BF build", "Saving gross (data reduced by the BFs)", "Saving net = No BF − Default"], ["ΔT vs default", "ΔC vs default"]] = ""
summ.round(2)
```

```python
# totals over the queries: No BF | Default (BF build part hatched) | Knee, for time and cost
fig, axs = plt.subplots(1, 2, figsize=(7.2, 2.9))
labels = ["No BF", "Default\n(BF always on)", "Knee\n(optimized)"]
cols = ["#8a8983", "#52514e", "#2a78d6"]
for ax, (key, unit, scale) in zip(axs, (("T", "s", 1), ("C", style.COST_UNIT, S))):
    v = [tot[f"{key}_no_bf{'_s' if key == 'T' else '_usd'}"] * scale, tot[f"{key}_default{'_s' if key == 'T' else '_usd'}"] * scale,
         tot[f"{key}_knee{'_s' if key == 'T' else '_usd'}"] * scale]
    bf = tot[f"{key}_bf_build{'_s' if key == 'T' else '_usd'}"] * scale
    bars = ax.bar(range(3), v, 0.62, color=cols, edgecolor="black", lw=0.6, zorder=2)
    ax.bar([1], [bf], 0.62, bottom=[v[1] - bf], color="none", edgecolor="white", hatch="////", lw=0, zorder=3)
    for i, b in enumerate(bars):
        txt = "" if i == 1 else f"{v[i] / v[1] - 1:+.0%}".replace("-", "−")
        ax.annotate(f"{v[i]:.1f}" + (f"\n{txt}" if txt else ""), (b.get_x() + b.get_width() / 2, v[i]),
                    xytext=(0, 3), textcoords="offset points", ha="center", va="bottom", fontsize=FONT_SIZE - 4)
    ax.set_xticks(range(3), labels, fontsize=FONT_SIZE - 3)
    ax.set_ylim(0, max(v) * 1.25)
    ax.set_ylabel(f"Total {'time' if key == 'T' else 'cost'} ({unit})")
    ax.grid(True, axis="y", color=style.GRID); ax.grid(False, axis="x"); ax.set_axisbelow(True)
from matplotlib.patches import Patch
fig.legend([Patch(facecolor="#52514e", edgecolor="white", hatch="////")], ["BF build part of the default plan"],
           loc="upper center", frameon=False, fontsize=FONT_SIZE - 3, bbox_to_anchor=(0.5, 1.04))
fig.tight_layout()
style.save(fig, FIG, "bf-tradeoff-totals")
plt.show()
```

## Pareto fronts

```python
from matplotlib.ticker import MaxNLocator
from matplotlib.lines import Line2D
GRAY, BLUE, ORANGE, GREEN = "#c9c8c2", "#2a78d6", "#eb6834", "#1baf7a"
PAD = 0.08


def panel(ax, q):
    R = PTS[PTS["query"] == q]
    F = R[R.pareto].sort_values("T")
    ax.scatter(R["T"], R["C"], s=5, color=GRAY, lw=0, rasterized=True)
    ax.plot(F["T"], F["C"], color=BLUE, lw=1.4, zorder=2)
    ax.scatter(F["T"], F["C"], s=14, color=BLUE, edgecolor="white", lw=0.6, zorder=3)
    k = PLANS[(q, "knee")]
    ax.scatter([k.T_est_s], [k.C_est_usd * style.COST_SCALE], s=95, facecolor="none", edgecolor="black", lw=1.2, zorder=4)
    for key, mk, col in (("min_T", "^", ORANGE), ("min_C", "s", GREEN)):
        p = PLANS[(q, key)]
        ax.scatter([p.T_est_s], [p.C_est_usd * style.COST_SCALE], marker=mk, s=30, color=col, edgecolor="white", lw=0.6, zorder=5)
    ax.set_title(q.upper(), fontsize=FONT_SIZE - 1)
    ax.grid(True, color=style.GRID, lw=0.6); ax.set_axisbelow(True)
    for vals, setlim in ((F["T"], ax.set_xlim), (F["C"], ax.set_ylim)):
        lo, hi = vals.min(), vals.max()
        pad = (hi - lo) * PAD or lo * 0.05
        setlim(lo - pad, hi + pad)
    ax.xaxis.set_major_locator(MaxNLocator(nbins=4, steps=[1, 2, 4, 5, 10], min_n_ticks=3))
    ax.yaxis.set_major_locator(MaxNLocator(nbins=5, steps=[1, 2, 2.5, 5, 10], min_n_ticks=3))


def figure(queries, name, ncol=3):
    nrow = -(-len(queries) // ncol)
    height = 1.9 * nrow + 0.5
    fig, axs = plt.subplots(nrow, ncol, figsize=(1.95 * ncol + 0.3, height), squeeze=False)
    for ax, q in zip(axs.flat, queries):
        panel(ax, q)
    for ax in list(axs.flat)[len(queries):]:
        ax.axis("off")
    for ax in axs[:, 0]:
        ax.set_ylabel(f"Estimated cost C ({style.COST_UNIT})")
    for c in range(ncol):
        used = [axs[r, c] for r in range(nrow) if r * ncol + c < len(queries)]
        if used:
            used[-1].set_xlabel("Estimated time T (s)")
    mk = lambda m, ms, fc, ec="white", mew=0.6: Line2D([], [], ls="", marker=m, ms=ms, mfc=fc, mec=ec, mew=mew)
    h = [mk("o", 4, GRAY, GRAY, 0), Line2D([], [], color=BLUE, lw=1.4, marker="o", ms=6, mfc=BLUE, mec="white", mew=0.6),
         mk("o", 11, "none", "black", 1.2), mk("^", 8, ORANGE), mk("s", 7, GREEN)]
    l = ["Evaluated plans", "Pareto front", "Selected plan (knee)", "Min-time plan", "Min-cost plan"]
    leg_ncol = 5 if ncol >= 4 else 3
    leg_rows = -(-len(l) // leg_ncol)
    fig.legend(h, l, loc="upper center", ncol=leg_ncol, frameon=False, bbox_to_anchor=(0.5, 1.0), fontsize=FONT_SIZE - 2,
               handlelength=1.6, handletextpad=0.4, columnspacing=1.0, labelspacing=0.3, borderpad=0.2)
    fig.tight_layout(pad=0.3, w_pad=0.8, h_pad=0.6, rect=(0, 0, 1, 1 - 0.26 * leg_rows / height))
    style.save(fig, FIG, name)
    plt.show()
```

```python
for i in range(0, len(QS), 3):
    group = QS[i:i + 3] if i + 3 <= len(QS) else QS[-3:]
    figure(group, "pareto-opt-" + "-".join(group))
```

# Part 2 — Is the predicted Pareto front right? (measured)

The predicted plans were exported as BLOOM-FaaS pipelines and run on the testbed: per query the default plan, the
knee, the minimum-time and minimum-cost plans, and points spread along the front (`front_01` = fastest …
`front_06` = cheapest; a point equal to another plan is run once, column `run_as`). `data/validation-runs.csv`
has one row per measured run (`T_s`, `C_usd`); only the queries already run are evaluated below.

How right the front is, per query and overall:
* **Error** of the predicted T and C of every plan (MAPE).
* **Order**: share of plan pairs that the model orders like the measurement, by T and by C.
* **Front kept**: share of the predicted front points that are still non-dominated once measured.
* **Direction vs. default**: share of plans for which the model gets right whether T (C) goes up or down
  compared with the default plan (only plans measured more than 5% away from default).

```python
VR = pd.read_csv(f"{DATA}/validation-runs.csv")
VR["C"] = VR["C_usd"] * style.COST_SCALE
MEAS = VR.groupby(["query", "plan"]).agg(runs=("T_s", "size"), T=("T_s", "median"), T_lo=("T_s", "min"), T_hi=("T_s", "max"),
                                         C=("C", "median"), C_lo=("C", "min"), C_hi=("C", "max")).reset_index()
V = PL.merge(MEAS, left_on=["query", "run_as"], right_on=["query", "plan"], suffixes=("", "_m")).drop(columns="plan_m")
V["T_est"], V["C_est"] = V["T_est_s"], V["C_est_usd"] * style.COST_SCALE
V["err_T_%"] = 100 * (V["T_est"] / V["T"] - 1)
V["err_C_%"] = 100 * (V["C_est"] / V["C"] - 1)
VQ = sorted(V["query"].unique(), key=qnum)
print(f"measured: {len(VQ)} of {PL['query'].nunique()} queries ({', '.join(q.upper() for q in VQ)}), "
      f"{len(V.drop_duplicates(['query', 'run_as']))} distinct plans run, {len(VR)} runs")
V[["query", "plan", "run_as", "T_est", "T", "err_T_%", "C_est", "C", "err_C_%", "runs"]].round(2)
```

```python
def accuracy(q):
    d = V[V["query"] == q].drop_duplicates("run_as")            # one row per distinct measured plan
    pairs = [(a, b) for i, a in enumerate(d.itertuples()) for b in list(d.itertuples())[i + 1:]]
    agree = lambda k, ke: np.mean([(getattr(a, ke) - getattr(b, ke)) * (getattr(a, k) - getattr(b, k)) > 0 for a, b in pairs]) if pairs else np.nan
    front = d[d.run_as.str.startswith(("front_", "knee", "min_"))]
    nd = [not any((o["T"] <= r["T"]) and (o["C"] <= r["C"]) and (o["T"] < r["T"] or o["C"] < r["C"]) for _, o in d.iterrows())
          for _, r in front.iterrows()]
    base = d[d.run_as == "default"].iloc[0] if (d.run_as == "default").any() else None
    dirs = {"T": [], "C": []}
    if base is not None:
        for _, r in d[d.run_as != "default"].iterrows():
            for m, me in (("T", "T_est"), ("C", "C_est")):
                if abs(r[m] / base[m] - 1) > 0.05:
                    dirs[m].append((r[m] > base[m]) == (r[me] > base[me]))
    return {"query": q.upper(), "plans": len(d), "MAPE T %": d["err_T_%"].abs().mean(), "MAPE C %": d["err_C_%"].abs().mean(),
            "order T %": 100 * agree("T", "T_est"), "order C %": 100 * agree("C", "C_est"),
            "front kept %": 100 * np.mean(nd) if nd else np.nan,
            "direction T %": 100 * np.mean(dirs["T"]) if dirs["T"] else np.nan,
            "direction C %": 100 * np.mean(dirs["C"]) if dirs["C"] else np.nan}


ACC = pd.DataFrame([accuracy(q) for q in VQ]).set_index("query")
ACC.loc["all"] = ACC.mean(numeric_only=True)
ACC.loc["all", "plans"] = ACC["plans"].iloc[:-1].sum()
ACC.round(1)


# knee vs. default and the end points, measured vs. predicted (sum over the measured queries)
rows = []
for p in ("knee", "min_T", "min_C"):
    qs_ok = [q for q in VQ if ((V["query"] == q) & (V.plan == p)).any() and ((V["query"] == q) & (V.plan == "default")).any()]
    a = V[V.plan == p].set_index("query").loc[qs_ok]; d = V[V.plan == "default"].set_index("query").loc[qs_ok]
    rows.append({"plan": p, "queries": len(qs_ok),
                 "ΔT predicted": f"{a['T_est'].sum() / d['T_est'].sum() - 1:+.0%}", "ΔT measured": f"{a['T'].sum() / d['T'].sum() - 1:+.0%}",
                 "ΔC predicted": f"{a['C_est'].sum() / d['C_est'].sum() - 1:+.0%}", "ΔC measured": f"{a['C'].sum() / d['C'].sum() - 1:+.0%}",
                 "faster than default (measured)": f"{int((a['T'] < d['T']).sum())}/{len(qs_ok)}",
                 "cheaper than default (measured)": f"{int((a['C'] < d['C']).sum())}/{len(qs_ok)}"})
print("vs. the default plan, totals over the queries:")
display(pd.DataFrame(rows).set_index("plan"))
missing = PL[["query", "run_as"]].drop_duplicates().merge(MEAS, left_on=["query", "run_as"], right_on=["query", "plan"], how="left")
print("plans that never completed a run:", [f"{q.upper()} {a}" for q, a, t in missing[["query", "run_as", "T"]].itertuples(index=False) if pd.isna(t)])
```

```python
# predicted (hollow) vs. measured (filled) plans, joined by a line; the predicted front in light blue
PLAN_MK = {"default": ("D", "#52514e"), "knee": ("o", "#2a78d6"), "min_T": ("^", "#eb6834"), "min_C": ("s", "#1baf7a")}


def validation_panel(ax, q):
    F = PTS[(PTS["query"] == q) & PTS.pareto].sort_values("T")
    ax.plot(F["T"], F["C"], color="#9ecae1", lw=1.2, zorder=1)
    d = V[V["query"] == q].drop_duplicates("run_as")   # a plan run once is drawn once (named plans first)
    for r in d.itertuples():
        mk, col = PLAN_MK.get(r.plan, ("o", "#7b3fa0"))
        ax.plot([r.T_est, r.T], [r.C_est, r.C], color=col, lw=0.7, alpha=0.7, zorder=2)
        ax.scatter([r.T_est], [r.C_est], marker=mk, s=34, facecolor="white", edgecolor=col, lw=1.0, zorder=3)
        ax.scatter([r.T], [r.C], marker=mk, s=34, color=col, edgecolor="white", lw=0.5, zorder=4)
    ax.set_title(q.upper(), fontsize=FONT_SIZE - 1)
    ax.grid(True, color=style.GRID, lw=0.6); ax.set_axisbelow(True)


def validation_figure(queries, name, ncol=3):
    ncol = min(ncol, len(queries))
    nrow = -(-len(queries) // ncol)
    single = len(queries) == 1
    w, h_ = (5.2, 2.8) if single else (2.6 * ncol + 0.4, 2.3 * nrow + 0.6)
    fig, axs = plt.subplots(nrow, ncol, figsize=(w, h_), squeeze=False)
    for ax, q in zip(axs.flat, queries):
        validation_panel(ax, q)
    for ax in list(axs.flat)[len(queries):]:
        ax.axis("off")
    for ax in axs[:, 0]:
        ax.set_ylabel(f"Cost C ({style.COST_UNIT})")
    for ax in axs[-1, :]:
        ax.set_xlabel("Time T (s)")
    mk = lambda m, fc, ec: Line2D([], [], ls="", marker=m, ms=7, mfc=fc, mec=ec, mew=1.0)
    hs = [Line2D([], [], color="#9ecae1", lw=1.2), mk("o", "white", "black"), mk("o", "black", "white")] + \
         [mk(m, c, "white") for m, c in PLAN_MK.values()] + [mk("o", "#7b3fa0", "white")]
    ls = ["Predicted front", "Predicted", "Measured", "Default", "Knee", "Min-time", "Min-cost", "Other front point"]
    if single:   # one panel: legend in a column on the right
        fig.legend(hs, ls, loc="center left", ncol=1, frameon=False, bbox_to_anchor=(0.66, 0.5),
                   fontsize=FONT_SIZE - 3, handlelength=1.4, handletextpad=0.3, labelspacing=0.35)
        fig.tight_layout(rect=(0, 0, 0.66, 1))
    else:
        rows = -(-len(ls) // 4)
        fig.legend(hs, ls, loc="upper center", ncol=4, frameon=False, bbox_to_anchor=(0.5, 1.0),
                   fontsize=FONT_SIZE - 3, handlelength=1.4, handletextpad=0.3, columnspacing=0.8, labelspacing=0.3)
        fig.tight_layout(rect=(0, 0, 1, 1 - 0.24 * rows / h_))
    style.save(fig, FIG, name)
    plt.show()
    plt.close(fig)


validation_figure(VQ, "pareto-validation")                 # all queries in one figure
```

```python
# one figure per query: figures/pareto-validation-<query>.{pdf,png}
for q in VQ:
    validation_figure([q], f"pareto-validation-{q}")
```
